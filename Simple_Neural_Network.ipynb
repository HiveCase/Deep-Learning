```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
import matplotlib.pyplot as plt
```

```python
torch.manual_seed(42)
```

```python
# Distances in miles for recent bike deliveries
distances = torch.tensor([[1.0], [2.0], [3.0], [4.0]], dtype=torch.float32)

# Corresponding delivery times in minutes
times = torch.tensor([[6.96], [12.11], [16.77], [22.21]], dtype=torch.float32)
```

```python
model = nn.Sequential(nn.Linear(1,1))
```

```python
loss_func = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.0001)
```

```python
# Training Loop
for i in range(10000):
  optimizer.zero_grad()
  outputs = model(distances)
  loss = loss_func(outputs, times)
  loss.backward()
  optimizer.step()
  if (i + 1) % 50 == 0:
        print(f"Epoch {i + 1}: Loss = {loss.item()}")
```

```python
# inference block
with torch.no_grad():
  test_distance = torch.tensor([[30.]], dtype=torch.float32)
  test_time = model(test_distance)
  print(f"The calculated time for {test_distance.item()} is: {round(test_time.item(),2)} minutes")
```
