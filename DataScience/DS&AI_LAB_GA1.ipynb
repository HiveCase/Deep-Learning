```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f(x):
    return x * np.sin(x) - np.exp(-x) / (x**2)

# Generate x values
x_vals = np.linspace(1, 5, 1000)
y_vals = f(x_vals)

# Plot
plt.figure(figsize=(8,6))
plt.plot(x_vals, y_vals, label='$f(x) = x \sin x - e^{-x}/x^2$')
plt.axhline(0, color='black', linestyle='--', linewidth=0.8)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()

# Find root numerically
from scipy.optimize import fsolve

# Initial guess
root_guess = 2.0
root = fsolve(f, root_guess)[0]
print(f"Root: {root:.2f}")
```

```python
import numpy as np

rng = np.random.default_rng(seed=42)
A = rng.integers(0, 100, (100, 200))
B = rng.integers(0, 100, (100, 300))

# Concatenate along axis 1 (columns)
C = np.concatenate((A, B), axis=1)

# C now has shape (100, 500)
# Find the 10th row (1-indexed, so index 9 in 0-indexing)
row_10 = C[9, :]  # 0-indexed, so row index 9 is the 10th row
mean_row_10 = np.mean(row_10)

print(f"{mean_row_10:.2f}")
```
